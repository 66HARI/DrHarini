### 6. Online and Offline orders restaurants percentage?

```python
if 'online_order' in df_zomato.columns:
    # Count the number of restaurants accepting online orders
    online_orders_count = df_zomato[df_zomato['online_order'] == 'Yes'].shape[0]

    # Count the number of restaurants not accepting online orders
    offline_orders_count = df_zomato[df_zomato['online_order'] == 'No'].shape[0]

    # Calculate percentages
    total_restaurants = len(df_zomato)
    online_orders_percentage = (online_orders_count / total_restaurants) * 100
    offline_orders_percentage = (offline_orders_count / total_restaurants) * 100

    # Print the results
    print(f"Percentage of restaurants accepting online orders: {online_orders_percentage:.2f}%")
```

### 7. What is the ratio b/w restaurants that provide and do not provide table booking.

```python
if 'book_table' in df_zomato.columns:
    # Count the number of restaurants providing table booking
    table_booking_yes_count = df_zomato[df_zomato['book_table'] == 'Yes'].shape[0]

    # Count the number of restaurants not providing table booking
    table_booking_no_count = df_zomato[df_zomato['book_table'] == 'No'].shape[0]

    # Calculate the ratio
    if table_booking_no_count != 0:
        ratio_table_booking = table_booking_yes_count / table_booking_no_count
    else:
        ratio_table_booking = float('inf')  # Handle division by zero gracefully

    # Print the results
    print(f"Ratio of restaurants providing table booking to those not providing: {ratio_table_booking:.2f}")
else:
    print("The dataset does not contain a column indicating table booking availability.")
```

### 8. Check for missing values in rating column. Drop the missing value from the column and convert the column to numeric type. e.g "3.5/5" convert it to 3.5.

```python
missing_values = df_zomato['rate'].isnull().sum()
print(f"Number of missing values in 'rate' column: {missing_values}")

# Drop rows with missing values in 'rate' column
df_zomato.dropna(subset=['rate'], inplace=True)

# Convert formats like "3.5/5" to numeric
df_zomato['rate'] = df_zomato['rate'].str.split('/').str[0]  # Take the first part before '/'

# Convert 'rate' column to numeric type
df_zomato['rate'] = pd.to_numeric(df_zomato['rate'], errors='coerce')

# Check conversion success
if pd.notnull(df_zomato['rate']).all():
    print("Conversion successful. Data type of 'rate' column is now numeric.")
else:
    print("Conversion failed. Some values in 'rate' column could not be converted to numeric.")

# Display first few rows of cleaned DataFrame
print(df_zomato.head())
```

### 9. Convert the approx_cost(for two people) in numeric type after removing the ","

```python
if 'approx_cost(for two people)' in df_zomato.columns:
    # Ensure the column is treated as strings
    df_zomato['approx_cost(for two people)'] = df_zomato['approx_cost(for two people)'].astype(str)
    
    # Remove commas and convert to numeric
    df_zomato['approx_cost(for two people)'] = pd.to_numeric(df_zomato['approx_cost(for two people)'].str.replace(',', ''), errors='coerce')
    
    # Check conversion success
    if pd.notnull(df_zomato['approx_cost(for two people)']).all():
        print("Conversion successful. Data type of 'approx_cost(for two people)' column is now numeric.")
    else:
        print("Conversion partially successful. Some values in 'approx_cost(for two people)' column could not be converted to numeric.")
        print("Rows with non-numeric values:")
        print(df_zomato[df_zomato['approx_cost(for two people)'].isnull()])
else:
    print("The dataset does not contain a column named 'approx_cost(for two people)'.")
```

### 10. Find the maximum, minimum and average "approx_cost(for two people)" in each  "listed_in(city)" for each "listed_in(type)"


```python
if 'listed_in(city)' in df_zomato.columns and 'listed_in(type)' in df_zomato.columns and 'approx_cost(for two people)' in df_zomato.columns:
    # Group by 'listed_in(city)' and 'listed_in(type)', then aggregate
    grouped_stats = df_zomato.groupby(['listed_in(city)', 'listed_in(type)'])['approx_cost(for two people)'].agg(['max', 'min', 'mean'])

    # Display the grouped statistics
    print("Maximum, Minimum, and Average 'approx_cost(for two people)' in each 'listed_in(city)' for each 'listed_in(type)':")
    print(grouped_stats)
else:
    print("The dataset is missing one or more of the required columns: 'listed_in(city)', 'listed_in(type)', or 'approx_cost(for two people)'.")
```

```python
def return_budget_restaurant(location, rest_type):
    # Filter restaurants based on location and restaurant type
    filtered_restaurants = df_zomato[(df_zomato['listed_in(location)'] == location) & 
                                     (df_zomato['rest_type'] == rest_type)]
    
    # Convert 'rate' column to numeric
    filtered_restaurants['rate'] = pd.to_numeric(filtered_restaurants['rate'].str.replace('/5', ''), errors='coerce')
    
    # Apply budget criteria: price < 400 and rating > 4
    budget_restaurants = filtered_restaurants[(filtered_restaurants['approx_cost(for two people)'] < 400) &
                                              (filtered_restaurants['rate'] > 4.0)]
    
    # Return the filtered dataframe
    return budget_restaurants[['name', 'approx_cost(for two people)', 'rate']]

# Call the function with location = 'Banashankari' and rest_type = 'Buffet'
budget_restaurants = return_budget_restaurant('Banashankari', 'Buffet')

# Print the results
print("Budget Restaurants in Banashankari with Buffet type:")
print(budget_restaurants[['name', 'approx_cost(for two people)', 'rate']])
```

### 13.Which are the most common cuisines in each locations.

```python
if 'location' in df_zomato.columns:
    # Group by location and count occurrences of each cuisine
    common_cuisines = df_zomato.groupby('location')['cuisines'].value_counts()

    # Print the most common cuisines in each location
    for location, cuisine_counts in common_cuisines.groupby(level=0):
        most_common_cuisine = cuisine_counts.idxmax()
        print(f"Most common cuisine in {location}: {most_common_cuisine[1]}")
else:
    print("The dataset does not contain a column indicating locations (e.g., 'Location' or 'City').")
```
